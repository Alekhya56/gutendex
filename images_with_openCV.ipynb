Blending images

```
import cv2
img1 = cv2.imread('/content/drive/My Drive/Colab_Notebooks/openCv_images/00-puppy.jpg')
img1= cv2.cvtColor(img1,cv2.COLOR_BGR2RGB)
img2 = cv2.imread('/content/drive/My Drive/Colab_Notebooks/openCv_images/watermark_no_copy.png')
img2 = cv2.cvtColor(img2,cv2.COLOR_BGR2RGB)
print(img1.shape)
print(img2.shape)
```

```
img1 = cv2.resize(img1,(1200,1200))
img2 = cv2.resize(img2,(1200,1200))
import matplotlib.pyplot as plt
plt.imshow(img1)
```

```
plt.imshow(img2)
```

```
blended = cv2.addWeighted(src1=img1,alpha=0.9,src2=img2,beta=0.1,gamma=0)
plt.imshow(blended)
```

blurring and smoothing

```
import cv2
img1 = cv2.imread('/content/drive/My Drive/Colab_Notebooks/openCv_images/00-puppy.jpg')
img1= cv2.cvtColor(img1,cv2.COLOR_BGR2RGB)
gamma = 1 # if gamma is below 1, it becomes lighter color(whitens), it aboce 1, it darker(balckens)
i = np.power(img1,gamma)
plt.imshow(i)
```

```
font = cv2.FONT_HERSHEY_COMPLEX
text =cv2.putText(img1,text='puppy',org=(200,200),fontFace=font,fontScale=12,color=(255,255,255))
# plt.imshow(text)
from google.colab.patches import cv2_imshow
cv2_imshow(text)
```

```
kernel = np.ones(shape=(7,7),dtype=np.float32)/25
dst = cv2.filter2D(img1,-1,kernel) # to make it blur by creating our own kernel
from google.colab.patches import cv2_imshow
cv2_imshow(dst)
```

```
blured = cv2.blur(img1,ksize=(5,5)) #inbuilt method for blur by openCV
```

# Working with Videos

```
import cv2
cap = cv2.VideoCapture(0)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
writer = cv2.VideoWriter('path/to/myvideo.mp4',cv2.VideoWriter_fourcc(*'XVID'),20,(width,height)) # for saving the video
while True:
  # continously capturing by frame by frame and coverting into gray
  ret, frame = cap.read()
  gray = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
  cv2.imshow('frame',gray)
  if cv2.waitKey(1) & 0xFF == ord('q'):
    break
cap.release()
cv2.destroyAllWindows()
```
